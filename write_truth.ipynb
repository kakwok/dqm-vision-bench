# Write truth

Human-authored ground truth for one image at a time: meta data, the DQM
shift instruction, the reference plot(s), and the input plot side-by-side.
Output filenames are computed automatically to match `evaluate.py`'s
`find_truth_file()` convention — you never need to type a path yourself, and
`tw.show()` always tells you the exact filename `save()` will write before
you write it. All logic lives in `truth_helpers.py`; this notebook just
drives it.

**Workflow**
1. Run the setup cell once (`tw = TruthWriter()`).
2. `tw.print_coverage()` — see which plots/runs still need truth.
3. Optionally `tw.set_plot_filter('PlotName')` to work through one plot only
   (`tw.set_plot_filter(None)` to go back to all plots).
4. `tw.show()` — meta data + truth filename, instruction, reference/input
   plots side by side. Re-run any time you want to redisplay it.
5. `tw.blank_answer()` — resets `tw.quote` / `tw.describe` / `tw.compare` /
   `tw.decision` for the current item.
6. Edit `tw.quote` / `tw.describe` / `tw.compare` / `tw.decision` in the next
   cell.
7. `tw.save()` — writes the truth file and advances to the next item needing
   truth. Use `tw.save(overwrite=True)` to replace an existing file, or
   `tw.save(advance=False)` to stay on the current item.
8. Navigate manually any time with `tw.next_item()`, `tw.prev_item()`,
   `tw.goto(i)`, or `tw.jump_to('PlotName', run_number)`.
9. `tw.only_missing(False)` to page back through already-annotated items too
   (e.g. to revise one); `tw.only_missing(True)` to return to just the gaps.

## Navigation

`tw.show()` re-renders the current item, in this order: meta data (plot,
run, event type, and the **exact truth filename it will write** on `save()`),
the instruction text, then the reference plot(s) and the input plot
side-by-side (input on the right) so you can compare them directly.

`tw.only_missing(True/False)` (default `True`) and `tw.set_plot_filter(name)`
(default `None`, meaning all plots) control which rows are in play; move
through them with `tw.next_item()`, `tw.prev_item()`, `tw.goto(i)`, or
`tw.jump_to('PlotName', run_number)`.

## Answer

`tw.blank_answer()` resets `tw.quote` / `tw.describe` / `tw.compare` /
`tw.decision` for whatever item `tw.show()` just displayed (pre-filling
`quote` with the event type). Edit them in the next cell, then run
`tw.save()`.

```python
tw.blank_answer()
```

```python
tw.quote = tw.quote  # edit: quote the relevant instruction text
tw.describe = ""      # edit: what the input plot looks like
tw.compare = ""       # edit: how it compares to the instruction/reference
tw.decision = "good"  # or "bad"
```

## Save

`tw.current_truth_path` (also printed at the top of `tw.show()`) is the
exact path this will write, in the same 4-section format as the existing
files in `truth/`. Refuses to overwrite an existing file unless you pass
`overwrite=True`; auto-advances to the next item unless you pass
`advance=False`.

```python
print('About to write:', tw.current_truth_path)
tw.save()
```
